```python
# Unzip and read the data
filenames = ['force_plate', 'joint_angles', 'joint_velos', 'landmarks']
dfs = {}

for filename in filenames:
    with zipfile.ZipFile(f'../data/data/full_sig/{filename}.zip', 'r') as z:
        with z.open(f'{filename}.csv') as f:
            dfs[filename] = pd.read_csv(f)

force_plate_df = dfs['force_plate']
joint_angles_df = dfs['joint_angles']
joint_velos_df = dfs['joint_velos']
landmarks_df = dfs['landmarks']
```

```python
# Merge joint angles with joint velocities
merged_df = joint_angles_df.merge(joint_velos_df, on=['time', 'session_swing'], suffixes=('', '_dup'))
merged_df = merged_df.loc[:, ~merged_df.columns.str.endswith('_dup')]

# Merge with landmarks
merged_df = merged_df.merge(landmarks_df, on=['time', 'session_swing'], suffixes=('', '_dup'))
merged_df = merged_df.loc[:, ~merged_df.columns.str.endswith('_dup')]

# Merge with force plate data last
# merged_df = merged_df.merge(force_plate_df, on=['time', 'session_swing'], suffixes=('', '_dup'))
# merged_df = merged_df.loc[:, ~merged_df.columns.str.endswith('_dup')]
```

```python
# Left join with merged data
joined_df = eligible_swings_df.merge(merged_df, on=['session_swing'], how='left')

# Keep only 'session_swing' and 'bat_length_in' from eligible_swings_df, and all other columns from merged_df except duplicates
cols_to_keep = ['session_swing', 'bat_length_in'] + [col for col in merged_df.columns if col not in ['session_swing']]
joined_df = joined_df[cols_to_keep]
```

```python
# Find the episode length per session
def find_episode_length(session_swing):
    session_data = joined_df[joined_df['session_swing'] == session_swing]
    episode_length = session_data['time'].max() - session_data['time'].min()
    return episode_length

# Apply the function to each session_swing
episode_lengths = joined_df['session_swing'].unique()
episode_lengths_df = pd.DataFrame({
    'session_swing': episode_lengths,
    'episode_length': [find_episode_length(swing) for swing in episode_lengths]
})
```

```python
# Find number of states per swing session
def find_num_states(df, session_swing):
    session_data = df[df['session_swing'] == session_swing]
    num_states = session_data['time'].nunique()
    return num_states

# Apply the function to each session_swing
num_states = joined_df['session_swing'].unique()
num_states_df = pd.DataFrame({
    'session_swing': num_states,
    'num_states': [find_num_states(joined_df, swing) for swing in num_states]
})
```

```python
# Convert joint angles from degrees to radians
def joint_angle_ranges(df):
    joint_angle_columns = df.columns[df.columns.str.contains('angle')]
    # Convert to radians in-place
    df[joint_angle_columns] = np.deg2rad(df[joint_angle_columns])
    ranges = {}
    for col in joint_angle_columns:
        ranges[col] = (df[col].min(), df[col].max())
    return ranges

# Get ranges for joint angles (now in radians)
joint_angle_ranges_df = joint_angle_ranges(joined_df)
print("Joint Angle Ranges (radians):")
for joint, (min_val, max_val) in joint_angle_ranges_df.items():
    print(f"{joint}: Min = {min_val}, Max = {max_val}")
```

```python
# Reduce the frequency of the merged dataframe to 100 Hz rather than 357.14 Hz
def reduce_frequency(df, target_freq):
    # Calculate the time difference in seconds
    time_diff = 1 / target_freq
    
    # Create a new time column based on the target frequency
    df['new_time'] = (df['time'] // time_diff) * time_diff
    
    # Drop duplicates based on the new time column
    reduced_df = df.drop_duplicates(subset=['session_swing', 'new_time'])
    
    # Reset the index
    reduced_df.reset_index(drop=True, inplace=True)
    
    return reduced_df

# Reduce the frequency of the merged dataframe to 100 Hz
reduced_merged_df = reduce_frequency(joined_df, 100)

# View the reduced dataframe
print("Reduced Merged Data Shape:", reduced_merged_df.shape)
reduced_merged_df.head()
```

```python
# Find number of states per swing session in the reduced dataframe
reduced_num_states = reduced_merged_df['session_swing'].unique()
reduced_num_states_df = pd.DataFrame({
    'session_swing': reduced_num_states,
    'num_states': [find_num_states(reduced_merged_df, swing) for swing in reduced_num_states]
})

# Distribution of states in the reduced dataframe
sns.histplot(reduced_num_states_df['num_states'], bins=30, kde=True)
plt.title('Distribution of Number of States (Reduced Frequency)')
plt.xlabel('Number of States')
plt.ylabel('Frequency')
plt.show()
```

```python
def pad_or_truncate_episode_df(df, session_swing, H=200):
    episode = df[df['session_swing'] == session_swing].copy()
    L = len(episode)
    if L < H:
        pad_count = H - L
        pad_rows = episode.iloc[[-1]].copy()
        pad_rows = pd.concat([pad_rows]*pad_count, ignore_index=True)
        pad_rows['terminated'] = 1
        episode['terminated'] = 0
        episode = pd.concat([episode, pad_rows], ignore_index=True)
    else:
        episode = episode.iloc[:H].copy()
        episode['terminated'] = 0
    episode['phase'] = np.linspace(0, 1, H)
    episode = episode.reset_index(drop=True)
    return episode

H = 200  # desired number of states per episode

# Use pd.Series to store session_swing values
session_swings = pd.Series(reduced_merged_df['session_swing'].unique(), name='session_swing')

# Use pd.concat on a generator expression for DataFrame output
standardized_df = pd.concat(
    (pad_or_truncate_episode_df(reduced_merged_df, swing, H) for swing in session_swings),
    ignore_index=True
)

# View the standardized dataframe
print("Standardized Data Shape:", standardized_df.shape)

# View the distribution of the number of states in the standardized dataframe
sns.histplot(standardized_df['session_swing'].value_counts(), bins=30, kde=True)
plt.title('Distribution of Number of States in Standardized Data')
plt.xlabel('Number of States')
plt.ylabel('Frequency')
plt.show()
```

```python
import pybullet as p
import pybullet_data
import numpy as np
from scipy.optimize import minimize
import matplotlib.pyplot as plt
import time

# Connect to PyBullet in DIRECT mode and load the Panda robot
client_id = p.connect(p.DIRECT)
p.setAdditionalSearchPath(pybullet_data.getDataPath())
panda_id = p.loadURDF("franka_panda/panda.urdf", useFixedBase=True)
panda_start_pos = [0, 0, 0]
panda_start_orient = p.getQuaternionFromEuler([0, 0, 0])
p.resetBasePositionAndOrientation(panda_id, panda_start_pos, panda_start_orient)
p.resetBaseVelocity(panda_id, linearVelocity=[0, 0, 0], angularVelocity=[0, 0, 0])
p.stepSimulation()

# Get link positions for all joints
num_joints = p.getNumJoints(panda_id)
link_positions = []
for joint_idx in range(num_joints):
    joint_info = p.getJointInfo(panda_id, joint_idx)
    link_state = p.getLinkState(panda_id, joint_idx)
    link_positions.append((joint_info[12].decode('utf-8'), link_state[0]))

print("Panda robot link positions (name, position):")
for name, pos in link_positions:
    print(f"{name}: {pos}")

p.disconnect()
```

```python
from scipy.spatial.transform import Rotation as R

# --- INIT PYBULLET ---
physicsClient = p.connect(p.GUI)  # You can use p.DIRECT if you don't want GUI
p.setGravity(0, 0, -9.81)  # Z-up gravity

# Load plane + Panda robot
p.loadURDF("plane.urdf")
panda_start_pos = [0, 0, 0]
panda_start_orient = p.getQuaternionFromEuler([0, 0, 0])
panda_id = p.loadURDF("../urdf/panda_arm_bat.urdf",
                      panda_start_pos,
                      panda_start_orient,
                      useFixedBase=True)

# Load baseball URDF
baseball_id = p.loadURDF("../urdf/baseball.urdf",
                         basePosition=[0, -1.0, 0.7])

# --- DEFINE LINKS / JOINTS ---
shoulder_link_index = 1
elbow_link_index = 3
wrist_link_index = 6
bat_link_index = 7
panda_joint_indices = [0, 1, 2, 3, 4, 5, 6]  # 7 DOF arm

# --- WORLD SETTINGS ---
box_height = 0.2

# --- JOINT BOUNDS ---
panda_joint_bounds = []
for joint_id in panda_joint_indices:
    joint_info = p.getJointInfo(panda_id, joint_id)
    joint_type = joint_info[2]
    lower_limit = joint_info[8]
    upper_limit = joint_info[9]

    if joint_type in [p.JOINT_REVOLUTE, p.JOINT_PRISMATIC]:
        panda_joint_bounds.append((lower_limit, upper_limit))
    else:
        # For fixed joints, add dummy bound
        panda_joint_bounds.append((0.0, 0.0))

# --- GET ROBOT SHOULDER POS ---
robot_shoulder_pos = np.array(p.getLinkState(panda_id, shoulder_link_index)[0])

# --- GET FRAMES FOR FIRST SESSION_SWING ---
first_session = standardized_df['session_swing'].iloc[0]
session_df = standardized_df[standardized_df['session_swing'] == first_session].reset_index(drop=True)

# --- OPTIMIZATION LOOP PER FRAME ---
from matplotlib import pyplot as plt
import matplotlib.animation as animation

frames_rgb = []  # To collect RGB images for video

# Initialize q_init from neutral pose
q_init = np.array([p.getJointState(panda_id, j)[0] for j in panda_joint_indices])

for frame_idx, row in session_df.iterrows():
    print(f"Processing frame {frame_idx+1}/{len(session_df)}")
    
    # --- EXTRACT HUMAN JOINT POS ---
    lsjc = np.array([row['lsjc_x'], row['lsjc_y'], row['lsjc_z']])
    lejc = np.array([row['lejc_x'], row['lejc_y'], row['lejc_z']])
    lwjc = np.array([row['lwjc_x'], row['lwjc_y'], row['lwjc_z']])
    sweet_spot = np.array([row['sweet_spot_x'], row['sweet_spot_y'], row['sweet_spot_z']])
    
    # --- TARGET POSITIONS (RELATIVE TO SHOULDER) ---
    elbow_rel = lejc - lsjc
    wrist_rel = lwjc - lsjc
    sweet_spot_rel = sweet_spot - lsjc
    
    target_elbow_pos = robot_shoulder_pos + elbow_rel
    target_wrist_pos = robot_shoulder_pos + wrist_rel
    target_sweet_spot_pos = robot_shoulder_pos + sweet_spot_rel
    
    # --- LOSS FUNCTION (combined elbow, wrist, bat tip, with orientation) ---
    def loss_fn(q):
        for i, joint_id in enumerate(panda_joint_indices):
            p.resetJointState(panda_id, joint_id, q[i])

        p.applyExternalForce(objectUniqueId=baseball_id,
                    linkIndex=-1,  
                    forceObj=[0,0, 9.81 * 0.1], # note mass = 0.1
                    posObj=[0,0,0],
                    flags=p.LINK_FRAME)
                    
        p.stepSimulation()
        
        # Get state of elbow, wrist, and bat links
        elbow_pos_robot, elbow_orient_robot = p.getLinkState(panda_id, elbow_link_index)[:2]
        wrist_pos_robot, wrist_orient_robot = p.getLinkState(panda_id, wrist_link_index)[:2]
        bat_pos_robot, bat_orient_robot = p.getLinkState(panda_id, bat_link_index)[:2]
        
        w_elbow = 1.0
        w_wrist = 2.0
        w_bat = 5.0
        
        loss = 0.0
        loss += w_elbow * np.linalg.norm(elbow_pos_robot - target_elbow_pos)**2
        loss += w_wrist * np.linalg.norm(wrist_pos_robot - target_wrist_pos)**2
        loss += w_bat * np.linalg.norm(bat_pos_robot - target_sweet_spot_pos)**2
        
        return loss
    
    # --- RUN OPTIMIZATION ---
    result = minimize(loss_fn, q_init, bounds=panda_joint_bounds, options={'maxiter': 200})
    
    # --- APPLY SOLUTION ---
    q_solution = result.x
    for i, joint_id in enumerate(panda_joint_indices):
        p.setJointMotorControl2(
            bodyUniqueId=panda_id,
            jointIndex=joint_id,
            controlMode=p.POSITION_CONTROL,
            targetPosition=q_solution[i],
            force=50
        )
    
    # Let simulation settle
    for _ in range(20):
        p.applyExternalForce(objectUniqueId=baseball_id,
                    linkIndex=-1,  
                    forceObj=[0,0, 9.81 * 0.1], # note mass = 0.1
                    posObj=[0,0,0],
                    flags=p.LINK_FRAME)

        p.stepSimulation()
    
    # --- RENDER FRAME ---
    width, height, rgbImg, _, _ = p.getCameraImage(
        width=640,
        height=480,
        viewMatrix=p.computeViewMatrix(
            cameraEyePosition=[2, 2, 2],  # Zoomed out further
            cameraTargetPosition=[0, 0, 0.5],
            cameraUpVector=[0, 0, 1]
        ),
        projectionMatrix=p.computeProjectionMatrixFOV(
            fov=60,
            aspect=640/480,
            nearVal=0.1,
            farVal=6.0  # Increase farVal for more depth
        )
    )
    
    rgb_array = np.reshape(rgbImg, (height, width, 4))
    frames_rgb.append(rgb_array[:, :, :3])  # Drop alpha
    
    # Update q_init for next frame → warm start optimization
    q_init = q_solution

# --- CREATE ANIMATION ---
fig = plt.figure(figsize=(8, 6))
im = plt.imshow(frames_rgb[0])
plt.axis('off')

def update(frame):
    im.set_array(frame)
    return [im]

ani = animation.FuncAnimation(fig, update, frames=frames_rgb, interval=100)
plt.show()

# OPTIONAL: To save as video:
ani.save("panda_swing_bat.gif", fps=10)

p.disconnect()
```

```python
from scipy.optimize import minimize
import numpy as np
import pandas as pd
from tqdm import tqdm  # Optional: for progress bars

# --- INIT PYBULLET ---
physicsClient = p.connect(p.DIRECT)  # You can use p.DIRECT if you don't want GUI
p.setGravity(0, 0, -9.81)  # Z-up gravity

# Load plane + Panda robot
p.loadURDF("plane.urdf")
panda_start_pos = [0, 0, 0]
panda_start_orient = p.getQuaternionFromEuler([0, 0, 0])
panda_id = p.loadURDF("../urdf/panda_arm_bat.urdf",
                      panda_start_pos,
                      panda_start_orient,
                      useFixedBase=True)

# Load baseball URDF
baseball_id = p.loadURDF("../urdf/baseball.urdf",
                         basePosition=[0, -1.0, 0.7])

# --- DEFINE LINKS / JOINTS ---
shoulder_link_index = 1
elbow_link_index = 3
wrist_link_index = 6
panda_joint_indices = [0, 1, 2, 3, 4, 5, 6]  # 7 DOF arm
box_height = 0.2
bat_link_index = 7  # Replace this with your correct bat tip link index!

# --- JOINT BOUNDS ---
panda_joint_bounds = []
for joint_id in panda_joint_indices:
    joint_info = p.getJointInfo(panda_id, joint_id)
    joint_type = joint_info[2]
    lower_limit = joint_info[8]
    upper_limit = joint_info[9]

    if joint_type in [p.JOINT_REVOLUTE, p.JOINT_PRISMATIC]:
        panda_joint_bounds.append((lower_limit, upper_limit))
    else:
        # For fixed joints, add dummy bound
        panda_joint_bounds.append((0.0, 0.0))

# --- GET ROBOT SHOULDER POS ---
robot_shoulder_pos = np.array(p.getLinkState(panda_id, shoulder_link_index)[0])

# Final result structure
all_trajectories = []

# Process all sessions
for session in tqdm(standardized_df['session_swing'].unique()):
    session_df = standardized_df[standardized_df['session_swing'] == session].reset_index(drop=True)
    q_init = np.array([p.getJointState(panda_id, j)[0] for j in panda_joint_indices])
    
    session_joint_angles = []  # Store the angles per frame
    session_rewards = []       # Store rewards per frame
    session_terminals = []     # Store terminals per frame
    
    # Fixed ball parameters
    ball_pos_init = np.array([0.0, -1.0, 0.7])
    ball_vel = np.array([0.0, 0.0, 0.0])  # Currently stationary
    
    HIT_REWARD = 100.0
    DISTANCE_WEIGHT = -10.0
    BAT_SPEED_WEIGHT = 1.0
    BALL_VELOCITY_WEIGHT = 2.0
    NO_HIT_PENALTY = -50.0
    MAX_STEPS = 200
    GROUND_THRESHOLD = 0.05  # Ball Z below this → ground contact
    
    ball_hit = False
    ball_hit_position = None

    # Step count for max steps
    step_count = 0
    
    # Bat speed needs prev position
    bat_pos_robot_prev = None
    delta_t = 0.01  # Adjust this if your data has a different framerate
    
    for frame_idx, row in session_df.iterrows():
        # --- Extract joint positions ---
        lsjc = np.array([row['lsjc_x'], row['lsjc_y'], row['lsjc_z']])
        lejc = np.array([row['lejc_x'], row['lejc_y'], row['lejc_z']])
        lwjc = np.array([row['lwjc_x'], row['lwjc_y'], row['lwjc_z']])
        sweet_spot = np.array([row['sweet_spot_x'], row['sweet_spot_y'], row['sweet_spot_z']])
        
        # --- Target positions ---
        elbow_rel = lejc - lsjc
        wrist_rel = lwjc - lsjc
        sweet_spot_rel = sweet_spot - lsjc
        
        target_elbow_pos = robot_shoulder_pos + elbow_rel
        target_wrist_pos = robot_shoulder_pos + wrist_rel
        target_sweet_spot_pos = robot_shoulder_pos + sweet_spot_rel
        
        # --- Loss function ---
        def loss_fn(q):
            for i, joint_id in enumerate(panda_joint_indices):
                p.resetJointState(panda_id, joint_id, q[i])

            # Apply force to the baseball (keep it in place)
            p.applyExternalForce(objectUniqueId=baseball_id,
                                 linkIndex=-1,  
                                 forceObj=[0,0, 9.81 * 0.1],  # mass * g
                                 posObj=[0,0,0],
                                 flags=p.LINK_FRAME)
            
            p.stepSimulation()
            
            elbow_pos_robot = np.array(p.getLinkState(panda_id, elbow_link_index)[0])
            wrist_pos_robot = np.array(p.getLinkState(panda_id, wrist_link_index)[0])
            bat_pos_robot = np.array(p.getLinkState(panda_id, bat_link_index)[0])
            
            w_elbow = 1.0
            w_wrist = 2.0
            w_bat = 5.0
            
            loss = 0.0
            loss += w_elbow * np.linalg.norm(elbow_pos_robot - target_elbow_pos)**2
            loss += w_wrist * np.linalg.norm(wrist_pos_robot - target_wrist_pos)**2
            loss += w_bat * np.linalg.norm(bat_pos_robot - target_sweet_spot_pos)**2
            
            return loss
        
        # --- Run optimization ---
        result = minimize(loss_fn, q_init, bounds=panda_joint_bounds, options={'maxiter': 200})
        
        q_solution = result.x
        # Get bat and ball positions after optimization
        elbow_pos_robot = np.array(p.getLinkState(panda_id, elbow_link_index)[0])
        wrist_pos_robot = np.array(p.getLinkState(panda_id, wrist_link_index)[0])
        bat_pos_robot = np.array(p.getLinkState(panda_id, bat_link_index)[0])
        ball_pos, _ = p.getBasePositionAndOrientation(baseball_id)
        
        # Store joint angles, bat pos, ball pos, and ball_hit flag
        session_joint_angles.append(
            np.concatenate([
                q_solution.copy(), 
                bat_pos_robot, 
                ball_pos, 
                [float(ball_hit)]
            ])
        )
        
        # Warm start next frame
        q_init = q_solution
        
        # --- Compute bat speed ---
        if bat_pos_robot_prev is None:
            bat_speed = 0.0
        else:
            bat_speed = np.linalg.norm((bat_pos_robot - bat_pos_robot_prev) / delta_t)
        
        bat_pos_robot_prev = bat_pos_robot.copy()
        
        # --- Compute distance to ball ---
        distance = np.linalg.norm(bat_pos_robot - ball_pos_init)
        
        # --- Compute reward ---
        if not ball_hit:
            if distance < 0.05:
                ball_hit = True
                ball_hit_position = ball_pos_init.copy()
                reward = HIT_REWARD + BALL_VELOCITY_WEIGHT * np.linalg.norm(ball_vel) + BAT_SPEED_WEIGHT * bat_speed
            else:
                reward = DISTANCE_WEIGHT * distance + BAT_SPEED_WEIGHT * bat_speed
        else:
            # After ball hit — ignore distance, reward for ball horizontal distance and velocity
            ball_pos, _ = p.getBasePositionAndOrientation(baseball_id)
            ball_xy_distance = np.linalg.norm([ball_pos[0] - ball_pos_init[0], ball_pos[1] - ball_pos_init[1]])
            BALL_DISTANCE_REWARD_WEIGHT = 5.0
            reward = BALL_VELOCITY_WEIGHT * np.linalg.norm(ball_vel) + BALL_DISTANCE_REWARD_WEIGHT * ball_xy_distance
        
        # --- Terminal flag logic ---
        terminal = False
        # Only set terminal True if ball hits ground or last step or 200 steps
        if ball_hit:
            ball_pos, _ = p.getBasePositionAndOrientation(baseball_id)
            if ball_pos[2] < GROUND_THRESHOLD:
                terminal = True
        if frame_idx == len(session_df) - 1 or frame_idx >= MAX_STEPS - 1:
            terminal = True

        # --- Save reward and terminal flag ---
        session_rewards.append(reward)
        session_terminals.append(terminal)
        
        # --- Apply penalty if ball was never hit and this is the last frame ---
        if (not ball_hit) and (frame_idx == len(session_df) - 1):
            session_rewards[-1] += NO_HIT_PENALTY
            session_terminals[-1] = True
    
    # --- Convert to np.array ---
    session_joint_angles = np.array(session_joint_angles)  # (num_frames, num_joints+extra)

    # --- Compute actions (angle deltas between consecutive frames, only for joint angles) ---
    num_angles = len(q_init)
    session_actions = np.diff(session_joint_angles[:, :num_angles], axis=0)  # (num_frames-1, num_angles)

    # --- Save result ---
    trajectory_dict = {
        'session_swing': session,
        'observations': session_joint_angles,  # states
        'actions': session_actions,              # deltas
        'rewards': session_rewards,          # rewards
        'terminals': session_terminals      # terminals
    }
    all_trajectories.append(trajectory_dict)

p.disconnect()  # Disconnect after each session to free resources
```
