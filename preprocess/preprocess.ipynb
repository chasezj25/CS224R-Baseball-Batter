```python
# Imports
import pandas as pd
import numpy as np
import zipfile
```

```python
# Unzip and read the data
filenames = ['force_plate', 'joint_angles', 'joint_velos', 'landmarks']
dfs = {}

for filename in filenames:
    with zipfile.ZipFile(f'../data/data/full_sig/{filename}.zip', 'r') as z:
        with z.open(f'{filename}.csv') as f:
            dfs[filename] = pd.read_csv(f)

force_plate_df = dfs['force_plate']
joint_angles_df = dfs['joint_angles']
joint_velos_df = dfs['joint_velos']
landmarks_df = dfs['landmarks']
```

```python
# View force plate data
print("Force Plate Data Shape:", force_plate_df.shape)
force_plate_df.head()
```

```python
# View joint angle data
print("Joint Angles Data Shape:", joint_angles_df.shape)
joint_angles_df.head()
```

```python
# View joint velocity data
print("Joint Velocities Data Shape:", joint_velos_df.shape)
joint_velos_df.head()
```

```python
# View landmarks data
print("Landmarks Data Shape:", landmarks_df.shape)
landmarks_df.head()
```

```python
# Merge joint angles with joint velocities
merged_df = joint_angles_df.merge(joint_velos_df, on=['time', 'session_swing'], suffixes=('', '_dup'))
merged_df = merged_df.loc[:, ~merged_df.columns.str.endswith('_dup')]

# Merge with landmarks
merged_df = merged_df.merge(landmarks_df, on=['time', 'session_swing'], suffixes=('', '_dup'))
merged_df = merged_df.loc[:, ~merged_df.columns.str.endswith('_dup')]

# Merge with force plate data last
# merged_df = merged_df.merge(force_plate_df, on=['time', 'session_swing'], suffixes=('', '_dup'))
# merged_df = merged_df.loc[:, ~merged_df.columns.str.endswith('_dup')]
```

```python
# View merged data
print("Merged Data Shape:", merged_df.shape)
merged_df.head()
```

```python
# Collect eligible swings
eligible_swings_df = pd.read_csv('../eligible_swings.csv')
eligible_swings_df.head()
```

```python
# Left join with merged data
merged_df = eligible_swings_df.merge(merged_df, on=['session_swing'], how='left')

# Drop all columns from eligible_swings_df except for 'session_swing'
merged_df = merged_df.loc[:, ~merged_df.columns.isin(eligible_swings_df.columns[1:])]
```

```python
# View the merged dataframe
print("Final Merged Data Shape:", merged_df.shape)
merged_df.head()
```

```python
# Find the episode length per session
def find_episode_length(session_swing):
    session_data = merged_df[merged_df['session_swing'] == session_swing]
    episode_length = session_data['time'].max() - session_data['time'].min()
    return episode_length

# Apply the function to each session_swing
episode_lengths = merged_df['session_swing'].unique()
episode_lengths_df = pd.DataFrame({
    'session_swing': episode_lengths,
    'episode_length': [find_episode_length(swing) for swing in episode_lengths]
})
```

```python
# View episode lengths
print("Episode Lengths Data Shape:", episode_lengths_df.shape)
episode_lengths_df.head()
```

```python
# Distribution of episode lengths
import matplotlib.pyplot as plt
import seaborn as sns

sns.histplot(episode_lengths_df['episode_length'], bins=30, kde=True)
plt.title('Distribution of Episode Lengths')
plt.xlabel('Episode Length')
plt.ylabel('Frequency')
plt.show()
```

```python
# Find number of states per swing session
def find_num_states(df, session_swing):
    session_data = df[df['session_swing'] == session_swing]
    num_states = session_data['time'].nunique()
    return num_states

# Apply the function to each session_swing
num_states = merged_df['session_swing'].unique()
num_states_df = pd.DataFrame({
    'session_swing': num_states,
    'num_states': [find_num_states(merged_df, swing) for swing in num_states]
})
```

```python
# View number of states
print("Number of States Data Shape:", num_states_df.shape)
num_states_df.head()
```

```python
# Distribution of states
sns.histplot(num_states_df['num_states'], bins=30, kde=True)
plt.title('Distribution of Number of States')
plt.xlabel('Number of States')
plt.ylabel('Frequency')
plt.show()
```

```python
# Observe ranges of joint angles for all joints
def joint_angle_ranges(df):
    joint_angle_columns = df.columns[df.columns.str.contains('angle')]
    ranges = {}
    for col in joint_angle_columns:
        ranges[col] = (df[col].min(), df[col].max())
    return ranges

# Get ranges for joint angles
joint_angle_ranges_df = joint_angle_ranges(merged_df)
print("Joint Angle Ranges:")
for joint, (min_val, max_val) in joint_angle_ranges_df.items():
    print(f"{joint}: Min = {min_val}, Max = {max_val}")
```

```python
# Calculate time differences between consecutive states for each session
merged_df['time_diff'] = merged_df.groupby('session_swing')['time'].diff()

# View the first few rows to check the result
merged_df[['session_swing', 'time', 'time_diff']].head(10)

# Make a histogram of proportion of time differences
plt.figure(figsize=(10, 6))
sns.histplot(
    merged_df['time_diff'].dropna(),
    bins=30,
    kde=True,
    stat='probability'
)
plt.title('Histogram of Time Differences Between States')
plt.xlabel('Time Difference (seconds)')
plt.ylabel('Proportion')
plt.show()
```

```python
# Convert time_diff to Hz
time_diff = 0.0028

# Calculate the frequency in Hz
frequency_hz = 1 / time_diff
print(f"Frequency of data collection: {frequency_hz} Hz")
```

```python
# Reduce the frequency of the merged dataframe to 100 Hz rather than 357.14 Hz
def reduce_frequency(df, target_freq):
    # Calculate the time difference in seconds
    time_diff = 1 / target_freq
    
    # Create a new time column based on the target frequency
    df['new_time'] = (df['time'] // time_diff) * time_diff
    
    # Drop duplicates based on the new time column
    reduced_df = df.drop_duplicates(subset=['session_swing', 'new_time'])
    
    # Reset the index
    reduced_df.reset_index(drop=True, inplace=True)
    
    return reduced_df

# Reduce the frequency of the merged dataframe to 100 Hz
reduced_merged_df = reduce_frequency(merged_df, 100)

# View the reduced dataframe
print("Reduced Merged Data Shape:", reduced_merged_df.shape)
reduced_merged_df.head()
```

```python
# Find number of states per swing session in the reduced dataframe
reduced_num_states = reduced_merged_df['session_swing'].unique()
reduced_num_states_df = pd.DataFrame({
    'session_swing': reduced_num_states,
    'num_states': [find_num_states(reduced_merged_df, swing) for swing in reduced_num_states]
})

# Distribution of states in the reduced dataframe
sns.histplot(reduced_num_states_df['num_states'], bins=30, kde=True)
plt.title('Distribution of Number of States (Reduced Frequency)')
plt.xlabel('Number of States')
plt.ylabel('Frequency')
plt.show()
```

```python
def pad_or_truncate_episode_df(df, session_swing, H=200):
    episode = df[df['session_swing'] == session_swing].copy()
    L = len(episode)
    if L < H:
        pad_count = H - L
        pad_rows = episode.iloc[[-1]].copy()
        pad_rows = pd.concat([pad_rows]*pad_count, ignore_index=True)
        pad_rows['terminated'] = 1
        episode['terminated'] = 0
        episode = pd.concat([episode, pad_rows], ignore_index=True)
    else:
        episode = episode.iloc[:H].copy()
        episode['terminated'] = 0
    episode['phase'] = np.linspace(0, 1, H)
    episode = episode.reset_index(drop=True)
    return episode

H = 200  # desired number of states per episode

# Use pd.Series to store session_swing values
session_swings = pd.Series(reduced_merged_df['session_swing'].unique(), name='session_swing')

# Use pd.concat on a generator expression for DataFrame output
standardized_df = pd.concat(
    (pad_or_truncate_episode_df(reduced_merged_df, swing, H) for swing in session_swings),
    ignore_index=True
)

# View the standardized dataframe
print("Standardized Data Shape:", standardized_df.shape)

# View the distribution of the number of states in the standardized dataframe
sns.histplot(standardized_df['session_swing'].value_counts(), bins=30, kde=True)
plt.title('Distribution of Number of States in Standardized Data')
plt.xlabel('Number of States')
plt.ylabel('Frequency')
plt.show()
```

```python
# Remove 'time_diff' column from the standardized dataframe
standardized_df = standardized_df.drop(columns=['time_diff', 'new_time'])
standardized_df.head()
```

```python
# Switch y to be z and vice versa for all positions and joint angles
# Get all columns that have 'y' in their names
y_columns = standardized_df.columns[standardized_df.columns.str.contains('y')]

# Get all columns that have 'z' in their names
z_columns = standardized_df.columns[standardized_df.columns.str.contains('z')]

# Switch y and z columns
for y_col, z_col in zip(y_columns, z_columns):
    standardized_df[z_col], standardized_df[y_col] = standardized_df[y_col].copy(), standardized_df[z_col].copy()

standardized_df.head()
```

```python
# Save the merged dataframe to a CSV file
standardized_df.to_csv('../rh_swing_data.csv', index=False)
```
